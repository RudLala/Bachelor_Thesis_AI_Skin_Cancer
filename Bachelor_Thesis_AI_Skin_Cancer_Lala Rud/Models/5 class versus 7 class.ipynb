```python
import os
import pandas as pd
import matplotlib.pyplot as plt

RESULTS_DIR = "/home/jovyan/results"
DATA_DIR = "/home/jovyan/data"

metadata_path = os.path.join(DATA_DIR, "HAM10000_metadata.csv")

df = pd.read_csv(metadata_path)

class_counts = df["dx"].value_counts().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

plt.bar(class_counts.index, class_counts.values)

plt.title("HAM10000 Class Distribution")
plt.xlabel("Lesion Class")
plt.ylabel("Number of Images")

plt.tight_layout()

save_path = os.path.join(RESULTS_DIR, "figure_1_ham10000_class_distribution.png")
plt.savefig(save_path, dpi=300)

plt.show()

print("Saved to:", save_path)
```

```python
final_classes = ["akiec", "bcc", "bkl", "mel", "nv"]

df_5 = df[df["dx"].isin(final_classes)]

class_counts_5 = df_5["dx"].value_counts().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

plt.bar(class_counts_5.index, class_counts_5.values)

plt.title("Final 5-Class Dataset Distribution")
plt.xlabel("Lesion Class")
plt.ylabel("Number of Images")

plt.tight_layout()

save_path = os.path.join(RESULTS_DIR, "figure_1b_5class_distribution.png")
plt.savefig(save_path, dpi=300)

plt.show()

print("Saved to:", save_path)
```
